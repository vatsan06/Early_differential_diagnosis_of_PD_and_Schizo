```python
import os
import glob
import scipy.io as sio

# Define the path to your folder
folder_path = r"parkinsons\PD Conflict Task"
mat_files = glob.glob(os.path.join(folder_path, "*.mat"))

print(f"Found {len(mat_files)} .mat files.")

# Inspect the first file to understand the structure
if mat_files:
    first_file = mat_files[0]
    print(f"\nInspecting: {os.path.basename(first_file)}")
    
    try:
        # Load the MATLAB file
        mat_data = sio.loadmat(first_file)
        
        # Print out the variables stored inside
        print("Variables in this file:")
        for key in mat_data.keys():
            if not key.startswith('__'): # Ignore hidden MATLAB metadata keys
                data_shape = getattr(mat_data[key], 'shape', 'No shape')
                print(f" - {key}: {data_shape}")
                
    except NotImplementedError:
        print("This file is saved in MATLAB v7.3 format. You will need to use the 'h5py' library instead of scipy.io.")
```

```python
import os
import scipy.io as sio

# Path to the specific file you just inspected
file_path = r"parkinsons\PD Conflict Task\8010_Session_1_PDDys_CC.mat"

# Load the data
mat_data = sio.loadmat(file_path, squeeze_me=True, struct_as_record=False)

# Extract the EEG struct
# Because we used squeeze_me=True, we can access it directly as an object
eeg_struct = mat_data['EEG']

print("--- Unpacking the EEG Structure ---")

# 1. The Actual Brainwave Data
# Usually shaped as (channels, timepoints) or (channels, timepoints, epochs)
data = eeg_struct.data
print(f"Brainwave Data shape: {data.shape}")

# 2. Sampling Rate
srate = eeg_struct.srate
print(f"Sampling Rate: {srate} Hz")

# 3. Channel Information
chans = eeg_struct.chanlocs
print(f"Number of Channels: {len(chans)}")

# 4. Event Markers (The Conflict Task stimuli)
events = eeg_struct.event
print(f"Number of Events: {len(events)}")

# Let's peek at the first few events to see what the markers are named
print("\nFirst 5 Event Types:")
for i in range(min(5, len(events))):
    # Depending on the MATLAB version, event types might be strings or numbers
    print(f" - Event {i+1}: {events[i].type}")
```

```python
import os
import scipy.io as sio

# Path to the specific file you just inspected
file_path = r"schizo\Cost Conflict in Schizophrenia\CCunproc\CC_Beh_s101_P.mat"

import numpy as np

# Load the data
mat_data = sio.loadmat(file_path, squeeze_me=True, struct_as_record=False)
beh = mat_data['Beh']

print("--- Unpacking the Behavioral (Beh) Data ---")
print(f"Base data type: {type(beh)}\n")

# Scenario A: It's an array of 4 items (like 4 task blocks)
if isinstance(beh, np.ndarray):
    for i in range(len(beh)):
        print(f"--- Element {i+1} ---")
        item = beh[i]
        
        # If the element is a MATLAB struct
        if hasattr(item, '_fieldnames'):
            print(f"Fields inside this struct: {item._fieldnames}")
            # Print the shape of the first field as a hint
            first_field = item._fieldnames[0]
            print(f"Shape of '{first_field}': {np.shape(getattr(item, first_field))}\n")
        else:
            # If it's just raw data
            print(f"Data shape: {np.shape(item)}\n")

# Scenario B: It's a single struct with exactly 4 fields
elif hasattr(beh, '_fieldnames'):
    print(f"Fields inside Beh:")
    for field in beh._fieldnames:
        data = getattr(beh, field)
        print(f" - {field}: shape {np.shape(data)}")
```
